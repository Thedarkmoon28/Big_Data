# Actividad 3 - Estadística descriptiva

**Propósito:** Aplicar medidas estadísticas básicas a una variable poblacional.



```python
df_2006 = df[df['Year'] == 2006]
df_mex = df[(df["Entity"] == "Mexico") & (df["Year"] >= 2006) & (df["Annual change in population"])]
df_mex.head()
```

```python
df_2006_2020_mex = df [(df["Year"] >= 2006) & (df["Year"] <= 2020) & (df["Code"] == "MEX")]
df_2006_2020_mex
```

```python
df_extraccion_rango_mex = df_2006_2020_mex[["Entity","Year","Annual change in population"]]
df_extraccion_rango_mex
```

```python
df_extraccion_rango_mex.mean(numeric_only=True)
```

```python
column = 'Annual change in population'
media = df_2006[column].mean()
mediana = df_2006[column].median()
varianza = df_2006[column].var()
desviacion = df_2006[column].std()
minimo = df_2006[column].min()
maximo = df_2006[column].max()

print(f"Resultados para el año 2006:")
print(f"Media: {media}")
print(f"Mediana: {mediana}")
print(f"Varianza: {varianza}")
print(f"Desviación estándar: {desviacion}")
print(f"Valor mínimo: {minimo}")
print(f"Valor máximo: {maximo}")
```

```python
entidad_min = df_2006[df_2006[column] == minimo]['Entity'].values[0]
entidad_max = df_2006[df_2006[column] == maximo]['Entity'].values[0]

print(f"El valor de la Media es: {media}")
print(f"El valor de la Mediana es: {mediana}")
print(f"La entidad con el valor mínimo es: {entidad_min}")
print(f"La entidad con el valor máximo es: {entidad_max}")
```

```python
# Resumen estadístico del año histórico seleccionado
year_selected = 2006
column = "Annual change in population"
df_2006 = df.loc[df["Year"].eq(year_selected)]
resumen_2006 = pd.Series({
    "Media": df_2006[column].mean(),
    "Mediana": df_2006[column].median(),
    "Varianza": df_2006[column].var(),
    "Desviación estándar": df_2006[column].std(),
    "Mínimo": df_2006[column].min(),
    "Máximo": df_2006[column].max(),
})
resumen_2006
```
